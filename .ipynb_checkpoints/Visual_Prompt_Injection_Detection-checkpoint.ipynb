## 2. Advanced AI: Computer Vision & NLP Preprocessing
Before training our models, let`s see how our Computer Vision (OCR) and Natural Language Processing (NLP) modules process a single malicious image.

```python
sample_img = "data/raw/malicious/malicious_0.jpg"
print(f"Analyzing Image: {sample_img}")

# 1. Computer Vision OCR Extraction
ocr_result = extract_text_from_image(sample_img)
print("\n--- OCR Extraction Result ---")
for key, value in ocr_result.items():
    print(f"{key}: {value}")

# 2. NLP Normalization
normalized = normalize_text(ocr_result["extracted_text"])
print("\n--- NLP Normalized Text ---")
print(normalized)
```
